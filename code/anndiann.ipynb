## import report.tsv

```python
def load_report(path, search_engine='fp'):
    """
    Read in long-form precursors file
    If search_engine is 'fp', read report.tsv
    If search_engine is 'bps', read results.tsv
    """
    if search_engine == 'fp':
        df = pd.read_csv(f'{path}/report.tsv', sep='\t')
    elif search_engine == 'bps':
        df = pd.read_csv(f'{path}/results.tsv', sep='\t')
    else:
        raise ValueError("Unsupported search engine")
    return df
```
