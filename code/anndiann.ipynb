```python
def load_report(path, search_engine='fp', header_only=False):
    """
    Read in long-form precursors file
    If search_engine is 'fp', read report.tsv
    If search_engine is 'bps', read results.tsv
    """
    file_name = 'report'
    if search_engine == 'bps':
        file_name = 'results'
    else:
        raise ValueError("Unsupported search engine")
    
    if header_only:
    ### load header 
        df = pd.read_csv(f'{path}/{file_name}.tsv', sep='\t', nrows=0)
    else:
        df = pd.read_csv(f'{path}/{file_name}.tsv', sep='\t')
    return df

#TODO: Add functionality to load partial report by level
#TODO: Use header only to predetermine columns to load
```
